```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ds=pd.read_csv('Homework2_Data.csv')
```

```python
ds
```

```python
dates=ds['Day']
tmin_forecast=['Tmin Forecast (°F)']
tmin_verification=['Tmin Verification (°F)']
```

```python
tmin_forecast
```

```python
# Configure the plot layout
plt.figure(figsize=(10, 6))
# Plot both Tmin datasets
plt.plot(
    dates,
    tmin_forecast,
    label="Tmin Forecast (°F)",
    color="#1f77b4",
    marker="o",
    linewidth=2,
    linestyle="--",
)
plt.plot(
    dates,
    tmin_verification,
    label="Tmin Verification (°F)",
    color="#ff7f0e",
    marker="s",
    linewidth=2,
)
# Add titles, labels, and formatting
plt.title(
    "Tmin Forecast vs. Tmin Verification Values For Dulles Airport, September 2026",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Temperature (°F)", fontsize=12)
# Add design elements for plot
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(fontsize=11, loc="upper right")
plt.tight_layout()
# Display the graph
plt.show()
```
